## Find path: from A to B

* A\* Search. 
* Heuristic function: manhattan distance

**❓Q: Which block is closer to B, C or D? (if consider manhattan distance as $h(n)$)**

<img src="images/heuristic_abcd.png" width="500px"> 
<img src="images/heuristic_manhattan.png" width="500px"> 

```python
import numpy as np
import heapq
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
```

```python

# ----------------------------
# 1) Environment (7x12)
# ----------------------------
H, W = 7, 12
A = (6, 0)
B = (0, 11)
C = (4, 1)
D = (2, 7)

gray_blocks = [
    (0,1),(0,3),(0,5),(0,6),(0,7),(0,10),
    (1,1),(1,3),(1,7),(1,9),(1,10),
    (2,3),(2,5),(2,9),(2,10),
    (3,0),(3,2),(3,3),(3,5),(3,7),(3,9),(3,10),
    (4,0),(4,5),(4,7),
    (5,0),(5,1),(5,2),(5,4),(5,5),
    (5,7),(5,8),(5,9),(5,10),(5,11),
    (6,4),(6,5)
]

grid = np.zeros((H, W), dtype=int)  # 0 free, 1 wall
for r, c in gray_blocks:
    grid[r, c] = 1

# Ensure special cells are free
for rc in [A, B, C, D]:
    grid[rc] = 0

def manhattan(p, q):
    return abs(p[0]-q[0]) + abs(p[1]-q[1])

def neighbors4(p):
    r, c = p
    for dr, dc in [(-1,0),(1,0),(0,-1),(0,1)]:
        rr, cc = r+dr, c+dc
        if 0 <= rr < H and 0 <= cc < W and grid[rr, cc] == 0:
            yield (rr, cc)
```

```python
# ----------------------------
# 2) A* Search (priority = f = g + h)
#    g(n) = path cost from A to n (each move cost 1)
# ----------------------------
def astar(start, goal):
    pq = []
    counter = 0

    g_score = {start: 0}
    f0 = manhattan(start, goal)
    heapq.heappush(pq, (f0, counter, start))
    counter += 1

    came_from = {start: None}
    closed = set()
    expanded_order = []   # order of node expansions for animation
    frames = []           # store per-step snapshots

    space_peak=len(pq)+len(closed)

    def snapshot(current, note=""):
        # frontier nodes for debugging/teaching (not required, but useful)
        frontier_nodes = [n for _,__,n in pq if n not in closed]
        frames.append({
            "current": current,
            "closed": set(closed),
            "g": dict(g_score),
            "came_from": dict(came_from),
            "frontier": set(frontier_nodes),
            "note": note
        })

    snapshot(start, "Initialize: push A with f=h(A) because g(A)=0")

    while pq:
        space_peak = max(space_peak, len(pq)+len(closed))
        f, _, node = heapq.heappop(pq)
        if node in closed:
            continue

        closed.add(node)
        expanded_order.append(node)
        snapshot(node, f"Pop node with min f. current={node}, f={f} (g={g_score[node]}, h={manhattan(node,goal)})")

        if node == goal:
            snapshot(node, "Goal reached!")
            break

        for nb in neighbors4(node):
            tentative_g = g_score[node] + 1  # each move cost 1
            if nb in closed and tentative_g >= g_score.get(nb, float("inf")):
                continue

            # If better path found or first time
            if tentative_g < g_score.get(nb, float("inf")):
                came_from[nb] = node
                g_score[nb] = tentative_g
                f_nb = tentative_g + manhattan(nb, goal)
                heapq.heappush(pq, (f_nb, counter, nb))
                counter += 1

        snapshot(node, "Expand current; relax neighbors (update g, push with f=g+h)")

    # reconstruct path
    path = []
    if goal in came_from:
        cur = goal
        while cur is not None:
            path.append(cur)
            cur = came_from[cur]
        path.reverse()
    
    print(f"peak memory={space_peak}")

    return expanded_order, path, frames

expanded_order, path, frames = astar(A, B)

print("A* result:")
print("Expanded nodes:", len(expanded_order))
print("Path length (cost):", (len(path)-1) if path else None)
```

```python
# ----------------------------
# 3) Visualization (show f = g + h on yellow cells)
# ----------------------------
def draw_cell(ax, r, c, face, edge="#cccccc", lw=0.8, z=1):
    ax.add_patch(Rectangle((c, r), 1, 1, facecolor=face, edgecolor=edge, linewidth=lw, zorder=z))

def draw_label(ax, pos, label, color):
    r, c = pos
    draw_cell(ax, r, c, color, z=10)
    ax.text(c+0.5, r+0.62, label, ha="center", va="center",
            fontsize=20, color="white", fontweight="bold", zorder=11)

def render_frame_astar(ax, frame_idx, show_h_on_black=True, show_path_after_goal=True):
    ax.clear()
    ax.set_aspect("equal")
    ax.set_xlim(0, W)
    ax.set_ylim(H, 0)
    ax.axis("off")

    ax.figure.patch.set_facecolor("black")
    ax.set_facecolor("black")

    f = frames[frame_idx]
    closed = f["closed"]
    g_map = f["g"]
    frontier = f["frontier"]

    # precompute h for all free cells
    h_map = {}
    for r in range(H):
        for c in range(W):
            if grid[r, c] == 0:
                h_map[(r, c)] = manhattan((r, c), B)

    # base grid
    for r in range(H):
        for c in range(W):
            if grid[r, c] == 1:
                face = "#555555"
            else:
                face = "#000000"
            draw_cell(ax, r, c, face)

            if show_h_on_black and grid[r, c] == 0 and (r, c) != B:
                ax.text(c+0.5, r+0.55, str(h_map[(r, c)]),
                        ha="center", va="center", fontsize=16,
                        color="white", fontweight="bold", zorder=2)

    # closed set (yellow) with f shown as g+h (like "1+16")
    for (r, c) in closed:
        if (r, c) in [A, B]:
            continue
        draw_cell(ax, r, c, "#ffd84d", z=3)
        g = g_map.get((r, c), None)
        h = h_map[(r, c)]
        if g is not None:
            ax.text(c+0.5, r+0.55, f"{g}+{h}",
                    ha="center", va="center", fontsize=14,
                    color="black", fontweight="bold", zorder=5)
    
    for (r, c) in frontier:
        if (r, c) in [A, B] or (r, c) in closed:
            continue
        draw_cell(ax, r, c, "#329FFF", z=3)
        g = g_map.get((r, c), None)
        h = h_map[(r, c)]
        if g is not None:
            ax.text(c+0.5, r+0.55, f"{g}+{h}",
                    ha="center", va="center", fontsize=14,
                    color="white", fontweight="bold", zorder=5)
    

    # current node outline
    current = f["current"]
    if current is not None:
        cr, cc = current
        ax.add_patch(Rectangle((cc, cr), 1, 1, fill=False, edgecolor="white", linewidth=3, zorder=12))

    # optional final path overlay once goal reached
    if show_path_after_goal and any("Goal reached!" in frames[i]["note"] for i in range(frame_idx+1)) and path:
        for (r, c) in path:
            if (r, c) not in [A, B]:
                ax.add_patch(Rectangle((c, r), 1, 1, facecolor="#6cda5f", edgecolor="#6cda5f", linewidth=0.0, zorder=4))
    # labels
    draw_label(ax, A, "A", "#b01515")
    draw_label(ax, B, "B", "#0b6a1f")
    # draw_label(ax, C, "C", "#0a72a8")
    # draw_label(ax, D, "D", "#0a72a8")

    # title
    ax.text(0.02, 1.08,
            f"A* Search (priority = f=g+h, h=Manhattan) | step {(frame_idx+1)//2}/{len(frames)//2}",
            transform=ax.transAxes, fontsize=18, color="black")
    ax.text(0.02, 1.03, frames[frame_idx]["note"], transform=ax.transAxes, fontsize=12, color="black")

# ----------------------------
# 4) Animate: one frame per second
# ----------------------------
fig, ax = plt.subplots(figsize=(11, 6))

def update(i):
    render_frame_astar(ax, i, show_h_on_black=True, show_path_after_goal=True)

anim = FuncAnimation(fig, update, frames=len(frames), interval=500, repeat=False)

plt.close(fig)
display(HTML(anim.to_jshtml()))
```

```python
# ----------------------------
# 2) IDA* Search
#    Iterative Deepening A*
#    g(n) = path cost from start
# ----------------------------
def idastar(start, goal):
    expanded_order = []   # record expansion order (across all iterations)
    frames = []           # store per-step snapshots

    # initial f-cutoff
    cutoff = manhattan(start, goal)

    # parent map (only meaningful along current DFS path / final solution)
    came_from = {start: None}

    # path set = implicit frontier (DFS stack)
    path_set = {start}

    stack = {start}
    max_space=len(stack)
    search_cost=0
    def snapshot(current, note=""):
        frames.append({
            "current": current,
            "path": list(path_set),   # DFS path = implicit frontier
            "stack": list(stack),
            "came_from": dict(came_from),
            "cutoff": cutoff,
            "note": note
        })

    snapshot(start, f"Initialize: cutoff = h(start) = {cutoff}")

    def successors_by_f(p, g, h_func):
        r, c = p
        result = []
        for dr, dc in [(-1,0),(1,0),(0,-1),(0,1)]:
            rr, cc = r+dr, c+dc
            if 0 <= rr < H and 0 <= cc < W and grid[rr, cc] == 0:
                new_state = (rr, cc)

                # compute f for this successor
                g_new = g + 1
                f = g_new + h_func(new_state, goal)
                result.append((new_state, f))

        # sort by f ascending
        result.sort(key=lambda x: x[1])
        # return state + action only (we keep sorted by f)
        return [state for state, _ in result]


    def dfs(node, g):
        nonlocal cutoff
        f = g + manhattan(node, goal)

        nonlocal max_space, search_cost, stack
        search_cost+=1

        # f-cutoff pruning
        if f > cutoff:
            snapshot(node, f"Prune: f={f} > cutoff={cutoff}")
            return f

        if node == goal:
            snapshot(node, "Goal reached!")
            return "FOUND"
        
        tobe_stack = successors_by_f(node, g, manhattan)
        for x in tobe_stack:
            stack.add(x) 
        max_space = max(max_space, len(stack))
        expanded_order.append(node)
        
        min_exceed = float("inf")

        for nb in tobe_stack:
            if nb in path_set:
                continue  # path-cycle check

            came_from[nb] = node
        snapshot(node, f"Expand node: f={f} (g={g}, h={manhattan(node,goal)})")

        for nb in tobe_stack:
            if nb in path_set:
                continue  # path-cycle check

            path_set.add(nb)

            t = dfs(nb, g + 1)

            if t == "FOUND":
                return "FOUND"

            if t < min_exceed:
                min_exceed = t

            # backtrack
            path_set.remove(nb)
            came_from.pop(nb, None)
            stack.remove(nb)

        return min_exceed

    # -------- main IDA* loop --------
    while True:
        t = dfs(start, g=0)

        if t == "FOUND":
            break

        if t == float("inf"):
            return (max_space, search_cost), [], frames  # no solution

        # raise cutoff to next f-contour
        snapshot(start, f"Increase cutoff: {cutoff} → {t}")
        cutoff = t

    # -------- reconstruct path --------
    path = []
    cur = goal
    while cur is not None:
        path.append(cur)
        cur = came_from[cur]
    path.reverse()

    return (max_space, search_cost), path, frames


H, W = 7, 12
A = (6, 0)
B = (0, 11)

gray_blocks = [
    (0,0),
    (1,0),(1,2),(1,3),(1,4),(1,5),
    (1,6),(1,7),(1,8),(1,9),(1,10),
    (2,0),(2,2),(2,10),
    (3,0),(3,2),(3,4),(3,5),
    (3,6),(3,7),(3,8),(3,10),
    (4,0),(4,4),(4,10),
    (5,0),(5,1),(5,2),(5,4),
    (5,6),(5,7),(5,8),(5,9),(5,10),
    (6,4)
]

grid = np.zeros((H, W), dtype=int)  # 0 free, 1 wall
for r, c in gray_blocks:
    grid[r, c] = 1

# Ensure special cells are free
for rc in [A, B]:
    grid[rc] = 0

(max_space, search_cost), path, frames = idastar(A, B)

print("IDA* result:")
print(f"search cost={search_cost}, memory cost={max_space}")
print("Path length (cost):", (len(path)-1) if path else None)

def _render_frame_idastar(ax, frame_idx, show_path_after_goal=True):
    ax.clear()
    ax.set_aspect("equal")
    ax.set_xlim(0, W)
    ax.set_ylim(H, 0)
    ax.axis("off")

    ax.figure.patch.set_facecolor("black")
    ax.set_facecolor("black")

    f = frames[frame_idx]
    current = f["current"]
    path_nodes = set(f.get("path", []))   # DFS path = implicit frontier
    cutoff = f.get("cutoff", None)
    note = f.get("note", "")
    stack = set(f.get("stack", []))

    # h values for all free cells
    h_map = {(r, c): manhattan((r, c), B) for r in range(H) for c in range(W) if grid[r, c] == 0}

    # base grid + h on free black cells
    for r in range(H):
        for c in range(W):
            face = "#555555" if grid[r, c] == 1 else "#000000"
            draw_cell(ax, r, c, face)

            if grid[r, c] == 0 and (r, c) != B:
                ax.text(c+0.5, r+0.55, str(h_map[(r, c)]),
                        ha="center", va="center", fontsize=16,
                        color="white", fontweight="bold", zorder=2)

    # highlight current DFS path in yellow, show "g+h" on those nodes
    # (in IDA*, g = depth from start along the current path)
    for (r, c) in path_nodes:
        if (r, c) in [A, B]:
            continue
        draw_cell(ax, r, c, "#ffd84d", z=3)

        # g = depth = len(path)-1 for that node if we can compute it
        # easiest: compute g by reconstructing along came_from map stored in the frame
        came_from = f.get("came_from", {})
        g = None
        if came_from and (r, c) in came_from:
            # walk back to start to get depth
            depth = 0
            cur = (r, c)
            seen = set()
            while cur is not None and cur not in seen:
                seen.add(cur)
                cur = came_from.get(cur, None)
                if cur is not None:
                    depth += 1
            g = depth

        h = h_map[(r, c)]
        if g is not None:
            ax.text(c+0.5, r+0.55, f"{g}+{h}",
                    ha="center", va="center", fontsize=14,
                    color="black", fontweight="bold", zorder=6)
            
    for (r, c) in stack:
        if (r, c) in [A, B]:
            continue
        draw_cell(ax, r, c, "#90cfff", z=2)

        came_from = f.get("came_from", {})
        g = None
        if came_from and (r, c) in came_from:
            # walk back to start to get depth
            depth = 0
            cur = (r, c)
            seen = set()
            while cur is not None and cur not in seen:
                seen.add(cur)
                cur = came_from.get(cur, None)
                if cur is not None:
                    depth += 1
            g = depth

        h = h_map[(r, c)]
        if g is not None:
            ax.text(c+0.5, r+0.55, f"{g}+{h}",
                    ha="center", va="center", fontsize=14,
                    color="black", fontweight="bold", zorder=6)

    # current outline
    if current is not None:
        cr, cc = current
        ax.add_patch(Rectangle((cc, cr), 1, 1, fill=False, edgecolor="black", linewidth=3, zorder=12))

    # overlay final path after goal reached (same as before)
    if show_path_after_goal:
        reached = any(("Goal reached!" in fr.get("note", "")) for fr in frames[:frame_idx+1])
        if reached and path:
            for (r, c) in path:
                if (r, c) not in [A, B]:
                    ax.add_patch(Rectangle((c, r), 1, 1,
                                           facecolor="#6cda5f", edgecolor="#6cda5f",
                                           linewidth=0.0, zorder=4))

    # labels
    draw_label(ax, A, "A", "#b01515")
    draw_label(ax, B, "B", "#0b6a1f")

    # header + note (use WHITE on black background)
    header = "IDA* Search (f=g+h, h=Manhattan)"
    if cutoff is not None:
        header += f" | cutoff={cutoff}"
    header += f" | frame {frame_idx+1}/{len(frames)}"

    ax.text(0.02, 1.08, header, transform=ax.transAxes, fontsize=18, color="black")
    ax.text(0.02, 1.03, note,   transform=ax.transAxes, fontsize=12, color="black")

fig, ax = plt.subplots(figsize=(11, 6))

def _update(i):
    _render_frame_idastar(ax, i, show_path_after_goal=True)

anim = FuncAnimation(fig, _update, frames=len(frames), interval=500, repeat=False)
plt.close(fig)
display(HTML(anim.to_jshtml()))
```
